```python
corr_mult=round((df_employee.pivot_table(values='annual_salary_usd', index='job_role', columns='education_level', aggfunc='mean')))

sns.heatmap(corr_mult, annot=True, cmap='Blues', fmt='.0f');

# default fmt = '.2g'
#  format fmt yang paling sering dan relevan digunakan dalam visualisasi data, beserta fungsinya:

# 1. Format Bilangan Bulat (Integer)
# Format ini hanya bisa digunakan jika matriks data Anda benar-benar bertipe integer (int), bukan float.

# 'd' : Decimal integer standar.
# Contoh output: 42, 1005
# ',d' : Integer dengan pemisah ribuan (menggunakan koma). Sangat berguna untuk angka yang besar.
# Contoh output: 1,000, 25,450

# 2. Format Bilangan Desimal (Float)
# Ini adalah format yang paling aman dan paling sering digunakan, terutama karena data numerik di Pandas/NumPy sering kali bertipe float. Angka di depan huruf 'f' menunjukkan jumlah digit di belakang koma.

# '.0f' : Float tanpa angka di belakang koma (dibulatkan). Alternatif terbaik jika data berupa float tapi ingin ditampilkan seperti integer.
# Contoh output: 42 (dari nilai asli 42.4)
# '.2f' : Float dengan 2 angka di belakang koma. Standar emas untuk matriks korelasi.
# Contoh output: 0.85, -3.14
# ',.2f' : Float dengan 2 angka di belakang koma DAN pemisah ribuan.
# Contoh output: 1,250.50

# 3. Format Persentase (Percentage)
# Format ini sangat berguna jika matriks berisi proporsi atau pecahan (misalnya dari 0.0 hingga 1.0) dan ingin menampilkannya sebagai persentase. Python akan secara otomatis mengalikan nilai tersebut dengan 100 dan menambahkan simbol %.

# '.0%' : Persentase tanpa desimal.
# Contoh output: 85% (dari nilai asli 0.85)
# '.1%' : Persentase dengan 1 angka di belakang koma.
# Contoh output: 12.5% (dari nilai asli 0.125)

# 4. Format Notasi Ilmiah (Scientific)
# Gunakan ini jika data memiliki rentang nilai yang sangat ekstrem (terlalu besar atau terlalu kecil/mendekati nol).

# '.2e' : Notasi eksponensial dengan 2 angka desimal.
# Contoh output: 1.25e+06 (berarti 1.25 juta)

# 5. Format Otomatis (General)
# 'g' atau '.4g' : General format. Python akan secara cerdas memilih apakah akan menampilkannya sebagai angka biasa atau notasi ilmiah, tergantung mana yang lebih mudah dibaca dan sesuai dengan ruang yang ada.
```
