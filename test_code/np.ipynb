```python
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
```

```python
gaji = [5000000, 12000000, 8500000]
nama = ['Andi', 'Budi', 'Siti']

fig, ax = plt.subplots()
ax.bar(nama, gaji)

# Versi LAMBDA: Sangat sempurna untuk kasus ini karena formatnya sederhana.
# FuncFormatter wajib menerima dua argumen: nilai (x) dan posisinya (pos).
ax.yaxis.set_major_formatter(FuncFormatter(lambda x, pos: f"{x/1e6:.1f} Juta"))

plt.title("Gaji Karyawan")
plt.show()
```

```python
import pandas as pd
import seaborn as sns

df_karyawan = pd.DataFrame({
    'Gaji': [5e6, 12e6, 8.5e6, 7e6, 20e6, 15e6],
    'Departemen': ['IT', 'HR', 'IT', 'Finance', 'Finance', 'IT'],
    'Umur': [25, 35, 28, 40, 45, 30]
})

# Versi DEF: Membuat fungsi otomasi visualisasi yang menghasilkan 2 plot sekaligus
def eda_kolom_numerik(dataframe, kolom_numerik, kolom_kategori):
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    
    # Plot 1: Distribusi keseluruhan
    sns.histplot(data=dataframe, x=kolom_numerik, kde=True, ax=axes[0])
    axes[0].set_title(f"Distribusi {kolom_numerik}")
    
    # Plot 2: Distribusi berdasarkan kategori
    sns.boxplot(data=dataframe, x=kolom_kategori, y=kolom_numerik, ax=axes[1])
    axes[1].set_title(f"{kolom_numerik} per {kolom_kategori}")
    
    plt.tight_layout()
    plt.show()

# Pemanggilan: Sangat bersih dan bisa diulang untuk kolom apapun
eda_kolom_numerik(df_karyawan, 'Gaji', 'Departemen')
eda_kolom_numerik(df_karyawan, 'Umur', 'Departemen')
```
